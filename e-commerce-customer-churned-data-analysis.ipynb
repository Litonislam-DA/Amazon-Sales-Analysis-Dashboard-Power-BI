```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
path = ('/kaggle/input/datasets/litonislam/e-commerce-customer-churn-cleaned-e-commerce-data/ecommerce_cleaned_final.csv')
df = pd.read_csv(path)
plt.figure(figsize=(12, 6))
sns.countplot(x='Country', hue='Churned', data=df, palette='viridis')
plt.title('Churn Distribution by Country', fontsize=16, fontweight='bold', color='#2c3e50')
plt.xlabel('Country', fontsize=12)
plt.ylabel('Count', fontsize=12)
plt.tight_layout(rect=[0, 0.03, 1, 0.95])
plt.savefig('Churn Distribution by Country.png', dpi=300)
plt.show()
```

```python

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
path = ('/kaggle/input/datasets/litonislam/e-commerce-customer-churn-cleaned-e-commerce-data/ecommerce_cleaned_final.csv')
df = pd.read_csv(path)
plt.figure(figsize=(10, 6))
sns.kdeplot(data=df, x='Age', hue='Churned', fill=True, palette='magma')
plt.title('Customer Age vs Churn Probability', fontsize=16, fontweight='bold', color='#2c3e50')
plt.tight_layout(rect=[0, 0.03, 1, 0.95])
plt.savefig('Customer Age vs Churn Probability.png', dpi=300)
plt.show()
```

```python

	
	

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

path = '/kaggle/input/datasets/litonislam/e-commerce-customer-churn-cleaned-e-commerce-data/ecommerce_cleaned_final.csv'
try:
    df = pd.read_csv(path)
    print("অভিনন্দন! ফাইলটি সফলভাবে লোড হয়েছে।")
    print(df.head())
except Exception as e:
    print("এরর হয়েছে:", e)
sns.set_theme(style="white")
plt.rcParams['figure.facecolor'] = '#f8f9fa'
fig, axes = plt.subplots(2, 2, figsize=(15, 15))
fig.suptitle('Professional E-commerce Customer Analysis Dashboard', fontsize=26, fontweight='bold', color='#1a1a1a',
             y=0.98)
churn_counts = df['Churned'].value_counts()
labels = ['Active', 'Churned']
colors = ['#2ecc71', '#e74c3c']
axes[0, 0].pie(churn_counts, labels=labels, autopct='%1.1f%%', startangle=140,
               colors=colors, wedgeprops={'width': 0.4, 'edgecolor': 'white', 'linewidth': 3},
               textprops={'fontsize': 14, 'fontweight': 'bold'})
axes[0, 0].set_title('Global Retention Status', fontsize=18, fontweight='bold', pad=20)
sns.countplot(ax=axes[0, 1], data=df, x='Country', hue='Churned', palette=['#1f77b4','#ff7f0e'], edgecolor='0.2')
axes[0, 1].set_title('Churn Distribution by Country', fontsize=18, fontweight='bold', pad=20)
axes[0, 1].tick_params(axis='x', rotation=30)
axes[0, 1].legend(title='Status', labels=['Active', 'Churned'])
sns.kdeplot(ax=axes[1, 0], data=df, x='Age', hue='Churned', fill=True, palette='magma', alpha=0.5, linewidth=2)
axes[1, 0].set_title('Customer Age vs Churn Probability', fontsize=18, fontweight='bold', pad=20)
sns.countplot(ax=axes[1, 1], data=df, x='Gender', hue='Churned', palette='husl', edgecolor='0.2')
axes[1, 1].set_title('Gender-Based Loyalty Metrics', fontsize=18, fontweight='bold', pad=20)
plt.tight_layout(rect=[0, 0.03, 1, 0.95])
plt.savefig('Professional_Dashboard_Output.png', dpi=300)
plt.show()
```
