```python


import numpy as np
import pandas as pd 
import seaborn as sns
import matplotlib.pyplot as plt
fill_path=('/kaggle/input/datasets/litonislam/superstore-sales-insights/Superstore Sales Dataset.csv')
df=pd.read_csv(fill_path)
plt.figure(figsize=(14, 8))
sns.set_style("white")
top_10_cities = df.groupby('City')['Sales'].sum().sort_values(ascending=False).head(10).reset_index()
ax = sns.barplot(x='Sales', y='City', data=top_10_cities,
                 palette='mako', edgecolor='black', linewidth=1.2)
plt.title('TOP 10 CITIES BY TOTAL REVENUE GENERATION',
          fontsize=22, fontweight='bold', color='#2C3E50', pad=40)
plt.xlabel('Total Revenue ($)', fontsize=14, fontweight='bold', labelpad=15)
plt.ylabel('City Name', fontsize=14, fontweight='bold')
plt.grid(axis='x', linestyle='--', alpha=0.5)
for i in ax.containers:
    ax.bar_label(i, padding=8, fontsize=12, fontweight='bold', fmt='$%.0f')
for spine in plt.gca().spines.values():
    spine.set_visible(True)
    spine.set_linewidth(1.5)
    spine.set_color('#333333')
    plt.savefig('Superstore Sales Insight Dataset',dpi=400)
plt.tight_layout()
plt.show()
```

Insights & Methodology

    ​Observation: The horizontal bar chart displays the "Top 10 Cities" by total revenue generation, providing a clear rank of market performance across different urban locations.
    ​Key Finding: New York City and Los Angeles lead the revenue generation, highlighting these cities as primary hubs for the business.
    ​How it’s done: Using the seaborn library, the data was grouped by city and aggregated to calculate total revenue. The chart was then plotted using a horizontal bar format with a gradient color palette to intuitively represent the revenue hierarchy.

```python

import numpy as np
import pandas as pd 
import seaborn as sns
import matplotlib.pyplot as plt
fill_path=('/kaggle/input/datasets/litonislam/superstore-sales-insights/Superstore Sales Dataset.csv')
df=pd.read_csv(fill_path)


segment_sales = df.groupby('Segment')['Sales'].sum()
plt.figure(figsize=(8, 8))
colors = ['#FF5733', '#33FF57', '#3357FF']
explode = (None)
plt.pie(segment_sales, labels=segment_sales.index, autopct='%1.1f%%',
        startangle=140, colors=colors, explode=explode, shadow=True,
        textprops={'fontsize': 13, 'fontweight': 'bold', 'color': 'black'},
        wedgeprops={'edgecolor': 'black', 'linewidth': 1.5})
plt.title('SALES DISTRIBUTION BY CUSTOMER SEGMENT',
          fontsize=20, fontweight='bold', color='#2C3E50', pad=30)
plt.savefig('Superstore Sales Insight Dataset',dpi=400)
plt.tight_layout()
plt.show()

```

Insights & Methodology

    ​Observation: The pie chart illustrates the proportional distribution of total sales across three key customer segments: Consumer, Corporate, and Home Office.

    ​How it’s done: I used the matplotlib.pyplot library to create this pie chart. The data was grouped by the 'Segment' column and aggregated by the sum of 'Sales' to calculate the percentage contribution for each segment, ensuring a clear visual breakdown of customer demographics

```python

import numpy as np
import pandas as pd 
import seaborn as sns
import matplotlib.pyplot as plt
fill_path=('/kaggle/input/datasets/litonislam/superstore-sales-insights/Superstore Sales Dataset.csv')
df=pd.read_csv(fill_path)
df['Order Date']=pd.to_datetime(df['Order Date'],dayfirst=True)
monthly_sales=df.resample('ME', on='Order Date')['Sales'].sum()
plt.figure(figsize=(15, 8))
sns.set_style("whitegrid")
plt.plot(monthly_sales.index, monthly_sales.values, color='#E67E22',
         marker='o', linewidth=3, markersize=8, markerfacecolor='#D35400', label='Monthly Revenue')
plt.fill_between(monthly_sales.index, monthly_sales.values, color='#E67E22', alpha=0.1)
plt.title('MONTHLY SALES PERFORMANCE TREND',
          fontsize=22, fontweight='bold', color='#D35400', pad=40)

plt.xlabel('Timeline (Year-Month)', fontsize=14, fontweight='bold', labelpad=15)
plt.ylabel('Total Revenue ($)', fontsize=14, fontweight='bold', labelpad=15)
plt.grid(True, linestyle='--', alpha=0.6)
plt.xticks(rotation=45, fontweight='bold')
plt.yticks(fontweight='bold')
for spine in plt.gca().spines.values():
    spine.set_visible(True)
    spine.set_linewidth(1.5)
plt.legend(fontsize=12)
plt.savefig('Superstore Sales Insight Dataset',dpi=400)
plt.tight_layout()
plt.show()
```

Insights & Methodology

    ​Observation: The line chart tracks the Monthly Sales Performance Trend over a four-year period (2015–2019), revealing the temporal fluctuations in revenue.
    ​Key Finding: The data shows clear seasonal volatility, with noticeable spikes and dips throughout each year, culminating in a significant upward growth trend toward the beginning of 2019.
    ​How it’s done: I used the matplotlib.pyplot library to plot the sales data against the timeline. The data was resampled to a monthly frequency, and a line plot was generated to visualize the continuous trend, providing a clear view of business seasonality and performance milestones.

Insights & Methodology

    ​Observation: The donut chart displays the percentage distribution of various Shipping Modes used for customer orders, providing insights into logistics preferences.
    ​Key Finding: Standard Class is the most preferred shipping method, accounting for the vast majority at 59.8%, which indicates it as the most cost-effective or preferred choice for customers.
    ​How it’s done: I used the matplotlib.pyplot library to generate this donut chart. The data was grouped by the 'Ship Mode' column, and counts were calculated to determine the proportional share of each shipping category, visualized with a centered circle to differentiate it from a standard pie chart.

```python
import numpy as np
import pandas as pd 
import seaborn as sns
import matplotlib.pyplot as plt
fill_path=('/kaggle/input/datasets/litonislam/superstore-sales-insights/Superstore Sales Dataset.csv')
df=pd.read_csv(fill_path)
plt.figure(figsize=(12, 8))
state_sales=df.groupby('State')['Sales'].sum().sort_values(ascending=False).head(15).reset_index()
sns.barplot(x='Sales', y='State', data=state_sales, palette='YlGnBu', edgecolor='black')
plt.title('TOP 15 STATES BY REVENUE GENERATION', fontsize=20, fontweight='bold', color='#145A32', pad=30)
plt.xlabel('Total Sales ($)', fontsize=14, fontweight='bold')
plt.ylabel('State', fontsize=14, fontweight='bold')
plt.grid(axis='x', linestyle=':', alpha=0.6)
plt.savefig('Superstore Sales Insight Dataset',dpi=400)
plt.tight_layout()
plt.show()
```

```python
import numpy as np
import pandas as pd 
import seaborn as sns
import matplotlib.pyplot as plt
fill_path=('/kaggle/input/datasets/litonislam/superstore-sales-insights/Superstore Sales Dataset.csv')
df=pd.read_csv(fill_path)
plt.figure(figsize=(12, 8))
sub_cat_sales = df.groupby('Sub-Category')['Sales'].sum().sort_values(ascending=True).reset_index()
plt.barh(sub_cat_sales['Sub-Category'], sub_cat_sales['Sales'], color='#8E44AD', edgecolor='black', alpha=0.8)
plt.title('SALES PERFORMANCE BY SUB-CATEGORY', fontsize=22, fontweight='bold', color='#4A235A', pad=30)
plt.xlabel('Revenue ($)', fontsize=14, fontweight='bold')
plt.ylabel('Sub-Category', fontsize=14, fontweight='bold')
plt.savefig('Superstore Sales Insight Dataset',dpi=400)
plt.tight_layout()
plt.show()
```

Insights & Methodology

    ​Observation: The horizontal bar chart provides a detailed breakdown of sales performance across various product Sub-Categories, ranking them from highest to lowest revenue generation.
    ​Key Finding: Phones and Chairs are the top-performing sub-categories, significantly leading in sales, while items like Fasteners and Labels represent the lower end of the sales volume.
    ​How it’s done: I used the matplotlib.pyplot library to visualize the aggregated sales data per sub-category. The chart was plotted horizontally to allow for clear comparison across the many product labels, using a consistent color theme for better readability and focus

```python
import numpy as np
import pandas as pd 
import seaborn as sns
import matplotlib.pyplot as plt
fill_path=('/kaggle/input/datasets/litonislam/superstore-sales-insights/Superstore Sales Dataset.csv')
df=pd.read_csv(fill_path)
total_sales = df['Sales'].sum()
plt.figure(figsize=(6,6))
plt.text(0.5, 0.6, 'TOTAL SALES', fontsize=14, fontweight='bold',
         color='#7F8C8D', ha='center')
plt.text(0.5, 0.3, f'${total_sales:,.2f}', fontsize=32, fontweight='bold',
         color='#2980B9', ha='center') # সেলস নীল রঙে
plt.box(on=True)
plt.axis('off')
plt.gca().add_patch(plt.Rectangle((0, 0), 1, 1, facecolor='#F4F6F7', edgecolor='#2980B9', linewidth=2))
plt.savefig('Superstore Sales Insight Dataset',dpi=400)
plt.tight_layout()
plt.show()
```

Insights & Methodology

    ​Observation: This indicator displays the Total Sales figure, providing an immediate snapshot of the overall revenue generated across the entire dataset.
    ​Key Finding: The total sales amount stands at $2,261,536.78, serving as the primary benchmark for measuring the superstore's overall financial performance.
    ​How it’s done: I calculated this value by summing the entire 'Sales' column in the dataset. This aggregate figure was then presented in a clean, simplified display to offer a quick high-level overview of business success before diving into specific category or regional breakdowns.
